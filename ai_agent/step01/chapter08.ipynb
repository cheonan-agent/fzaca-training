## 8. 코드를 나누고 가져다 쓰기 — 모듈·패키지·.env

```python
import datetime
import random
import math

now = datetime.datetime.now()
print(type(now))
print(now.strftime("%Y-%m-%d %H:%M"))

greetings = ["안녕하세요!", "반갑습니다!", "무엇을 도와드릴까요?"]
print(random.choice(greetings))

print(math.ceil(7 / 2))
```

파이썬에서 import 모듈명 as 약자는 모듈이나 패키지를 불러올 때 별칭(Alias)을 지정해 짧게 호출할 수 있게 해주는 문법입니다

```python
# step01/chapter08.py
import ai_agent.app.chat_utils as chat_utils

conversation = []
chat_utils.add_message(conversation, "system", "너는 사내 규정 도우미다.")
chat_utils.add_message(conversation, "user", "연차는 며칠이야?")
print(chat_utils.format_history(conversation, show_system=True))
print(chat_utils.ALLOWED_ROLES)
```

```python
import os
from dotenv import load_dotenv

load_dotenv()          # .env 파일을 읽어 환경 변수로 등록

api_key = os.getenv("OPENAI_API_KEY")
model = os.getenv("MODEL_NAME", "gpt-6-luna")
timeout = os.getenv("TIMEOUT_SECONDS", "30")

print("API 키 설정 여부:", bool(api_key))
print(f"모델: {model}")
print(f"타임아웃: {timeout} ({type(timeout).__name__})")
print(os.getenv("NOT_EXIST"))
```
