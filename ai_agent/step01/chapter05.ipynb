```python
pair = ("user", "연차는 며칠이야?")
role, content = pair          # 언패킹: 한 번에 나눠 담기

print(role)
print(content)
```

```python
allowed_roles = {"system", "user", "assistant"}
print("user" in allowed_roles)
print("admin" in allowed_roles)

tags = ["연차", "출장", "연차", "보안", "연차"]
print(set(tags))
print(len(set(tags)))
```
