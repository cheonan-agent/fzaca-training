### 5. 여러 데이터를 묶기 — 리스트·딕셔너리·튜플·세트

```python
questions = ["연차는 며칠이야?", "출장비 한도는?", "비밀번호 변경 주기는?"]

print(questions)
print(len(questions))
print(questions[0])
print(questions[-1])
print(questions[1:3])
print(questions[3])
```

```python
questions.append("교육비 지원되나?")    # 맨 뒤에 추가
print(questions)
print(questions[-2:])                     # 마지막 2개
print("출장비 한도는?" in questions)      # 들어 있는가

last = questions.pop()                    # 맨 뒤 것을 꺼내며 삭제
print(last)
print(questions)

print("출장비" in questions)  # 
```
