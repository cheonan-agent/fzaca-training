# 6. 걸러내고 반복하기 — 조건문·반복문·컴프리헨션

```python
conversation = [
    {"role": "system", "content": "너는 사내 규정 도우미다."},
    {"role": "user", "content": "연차는 며칠이야?"},
    {"role": "assistant", "content": "입사 1년 이상이면 연 15일이다."},
    {"role": "user", "content": "출장비 한도는?"},
    {"role": "assistant", "content": "식비 30,000원, 숙박비 100,000원이다."},
]

message = conversation[1]

if message["role"] == "user":
    print("사용자 질문이다")
```

```python
for message in conversation:
    print(message["role"], ":", message["content"])
```

```python
for i, message in enumerate(conversation):
    print(i, message["role"])
```

```python
for message in conversation:
    if message["role"] == "system":
        continue                      # 이번 차례는 건너뛴다
    if message["role"] == "user":
        print(f"👤 {message['content']}")
    else:
        print(f"🤖 {message['content']}")
```

```python
banned = ["비밀번호", "주민번호", "계좌"]
text = "내 비밀번호를 바꾸고 싶어"

found = False
for word in banned:
    if word in text:
        found = True
        break
print(found)
```

```python
print("앵무새 챗봇입니다. '종료'를 입력하면 끝납니다.")
conversation = []

while True:
    user_input = input("👤 ").strip()

    if user_input == "종료":
        print("대화를 종료합니다.")
        break
    if not user_input:
        continue

    conversation.append({"role": "user", "content": user_input})
    reply = f"'{user_input}'라고 하셨네요."
    conversation.append({"role": "assistant", "content": reply})
    print(f"🤖 {reply}")

print(f"총 메시지 수: {len(conversation)}")
```
