````python
print("앵무새 챗봇입니다. '종료'를 입력하면 끝납니다.")
conversation = []

while True:
    user_input = input("👤 ").strip()

    if user_input == "종료":
        print("대화를 종료합니다.")
        break
    if not user_input:
        continue

    conversation.append({"role": "user", "content": user_input})
    reply = f"'{user_input}'라고 하셨네요."
    conversation.append({"role": "assistant", "content": reply})
    print(f"🤖 {reply}")

print(f"총 메시지 수: {len(conversation)}")
````

## 4. 컴프리헨션 — 반복문 한 줄로 작성하기

"리스트를 돌면서 조건에 맞는 것만 모아 새 리스트 만들기"는 너무 자주 하는 일이라 한 줄 문법이 있다.

### 4-1. 같은 일을 두 방식으로

#### for 문 방식

```python
questions = []                      # ①  결과 담을 빈 리스트
for m in conversation:              # ②  하나씩 꺼내기
    if m["role"] == "user":         # ③  조건 검사
        questions.append(m["content"])   # ④  뽑아서 담을 값
```

#### 컴프리헨션 방식

```python
questions = [ m["content"]  for m in conversation  if m["role"] == "user" ]
#               ④ 뽑을 값        ② 하나씩 꺼내기        ③ 조건
```

```python
questions = [
    m["content"]
    for m in conversation
    if m["role"] == "user"
]
```

````python
questions = [
    m["content"]
    for m in conversation
    if m["role"] == "user"
]

print(questions)
````

````python
a = 1
b = a
b = 2
print(a)
print(id(a), id(b))

a = [1,2,3]
b = a
b[0] = 4
print(a)
print(id(a), id(b))
````

모르고 있었던 게 아니라, **아마 알고는 있었는데 정확한 모델로 정리가 안 돼 있었던** 걸 거예요. 이 부분은 파이썬 경력자도 자주 헷갈립니다.

우선 한 가지 정정하면, b = a는 "메모리 포인터 복사"라기보다 **"b라는 이름을 a가 가리키는 객체에 붙이는 것"**에 가깝습니다. 

**파이썬에서 변수는 값을 담는 상자가 아니라 객체에 붙은 이름표(name binding)예요.**

### 4-3. 하나라도 만족하는가 — `any()`

2-3절의 금칙어 검사를 한 줄로 줄일 수 있다. (`text`와 `banned`는 2-3에서 만든 것)

````python
banned = ["비밀번호", "주민번호", "계좌"]
text = "내 비번 바꾸고 싶음"

print(any(word in text for word in banned))
````

````python
counts = {}
for m in conversation:
    counts[m["role"]] = counts.get(m["role"], 0) + 1
print(counts)
````

````python
banned = ["비밀번호", "주민번호", "계좌"]
conversation = []

while True:
    user_input = input("👤 ").strip()
    if user_input == "종료":
        break
    if user_input == "기록":
        for m in conversation:
            print(f"   [{m['role']}] {m['content']}")
        continue
    if any(word in user_input for word in banned):
        print("🤖 민감한 정보는 입력하지 마세요.")
        continue
    conversation.append({"role": "user", "content": user_input})
    reply = f"'{user_input}'라고 하셨네요."
    conversation.append({"role": "assistant", "content": reply})
    print(f"🤖 {reply}")
````
