```python
import json

conversation = [
    {"role": "user", "content": "연차는 며칠이야?"},
    {"role": "assistant", "content": "입사 1년 이상이면 연 15일이다."},
]

s = json.dumps(conversation, ensure_ascii=False, indent=2)
print(type(s))
print(s)
```

```python
data = json.loads('{"role": "user", "content": "출장비 한도는?", "count": 3, "ok": true, "memo": null}')
print(type(data))
print(data)
```

```python
with open("../data/conversation.json", "w", encoding="utf-8") as f:
    json.dump(conversation, f, ensure_ascii=False, indent=2)

with open("../data/conversation.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)

print(type(loaded), len(loaded))
print(loaded[-1]["content"])
print(loaded == conversation)
```

**도전** — 6장에서 만든 앵무새 챗봇(**`step01/parrot.py`**)에 `app/storage.py`를 연결한다. 시작할 때 이전 대화를 불러오고, 메시지가 추가될 때마다 저장한다. 다시 실행해서 `기록`을 입력하면 이전 대화가 보여야 한다. 실행은 `uv run python -m step01.parrot`으로 한다.

```python
def search_regulations(keyword, path="../data/regulations.txt"):
    with open(path , "r", encoding="utf-8") as f:
        return [line.strip() for line in f if keyword in line]

for line in search_regulations("보안"):
    print(line)
```

```python
search_regulations("보안")
```

```python
sections = {}
current = None

with open("../data/regulations.txt", "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line.startswith("제") and "조 (" in line:
            current = line
            sections[current] = []
        elif line and current:
            sections[current].append(line)

for title, body in sections.items():
    print(title, len(body))
```
