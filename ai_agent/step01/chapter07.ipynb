```python
def num_adder(*args):
    t = 0
    for i in args:
        t = t+i
    return t

print(num_adder(1,2,3,4,5))
```

```python
def format_history(conversation, show_system=False):
    lines = []
    for m in conversation:
        if m["role"] == "system" and not show_system:
            continue
        lines.append(f"[{m['role']}] {m['content']}")
    return "\n".join(lines)

print(format_history(conversation))
print("---")
print(format_history(conversation, show_system=True))
```

```python
def count_by_role(conversation):
    "users와 assistant의 개수를 출력"
    user_count = 0
    assistant_count = 0
    for m in conversation:
        if m["role"] == "user":
            user_count += 1
        elif m["role"] == "assistant":
            assistant_count += 1
    return user_count, assistant_count

users, assistants = count_by_role(conversation)
print(users, assistants)
```

```python
def contains_banned(text, banned=("비밀번호", "주민번호", "계좌")):
    """금칙어가 하나라도 있으면 True를 돌려준다."""
    return any(word in text for word in banned)

print(contains_banned("비밀번호 바꾸는 법"))
print(contains_banned("연차 신청 방법"))
print(contains_banned("메신저 전송", banned=["메신저"]))
```

```python
# 기본
ALLOWED_ROLES = {"system", "user", "assistant"}

def is_valid_message(message):
    content = message.get("content")
    return message.get("role") in ALLOWED_ROLES and isinstance(content, str) and bool(content.strip())

print(is_valid_message({"role": "user", "content": "연차?"}))     # True
print(is_valid_message({"role": "admin", "content": "연차?"}))    # False
print(is_valid_message({"role": "user", "content": "   "}))        # False
```

```python
# 확장
def trim_history(conversation, max_messages=2):
    system = [m for m in conversation if m["role"] == "system"]
    others = [m for m in conversation if m["role"] != "system"]
    return system + others[-max_messages:]

for m in trim_history(conversation):
    print(m["role"], m["content"])
```
