# 12. 데이터와 기능을 묶기 — 클래스와 데코레이터

```python
class Conversation:
    def __init__(self, session_id):
        self.session_id = session_id
        self.messages = []

    def add (self, role, content):
        self.messages.append({"role": role, "content": content})

a = Conversation("user01")
b = Conversation("user02")

print(a.session_id, a.messages)
print(b.session_id, b.messages)
print(type(a))
```

## 5. 데코레이터 - 함수에 기능 덧입히기

```python
def greet():
    return "안녕하세요"

say = greet()
print(say)

def shout(func):
    def wrapper():
        return func() + "!!!"
    return wrapper

loud_greet = shout(greet)
print(loud_greet())
```

- `shout`는 함수를 받아서, 원래 함수를 감싼 새 함수(`wrapper`)를 돌려준다. 이것이 **데코레이터(decorator)**의 정체다.
- `greet = shout(greet)`를 짧게 쓰는 문법이 함수 위에 붙이는 `@shout`다.

## 5-2 실행 시간 재기 - @timer

```python
import functools
import time

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"[timer] {func.__name__} 실행시간: {elapsed:.2f}초")
        return result
    return wrapper


@timer
def slow_answer(question):
    time.sleep(0.5)
    return f"'{question}'에 대한 답변"

print(slow_answer("연차는 며칠이야?"))
print(slow_answer.__name__)
```
