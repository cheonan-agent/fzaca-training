# 12. 데이터와 기능을 묶기 — 클래스와 데코레이터

```python
class Conversation:
    def __init__(self, session_id):
        self.session_id = session_id
        self.messages = []

    def add(self, role, content):
        self.messages.append({"role": role, "content": content})

a = Conversation("user01")
b = Conversation("user02")
a.add("user", "연차는 며칠이야?")

print(a.session_id, a.messages)
print(b.session_id, b.messages)
print(type(a))
```
